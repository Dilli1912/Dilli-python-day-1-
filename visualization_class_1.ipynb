###Sales Trend Over time

```python
sales_by_date=D1.groupby('Order Date')['Sales'].sum()
sales_by_date.head()
```

```python
sales_by_date.index
```

```python
plt.plot(sales_by_date.index,sales_by_date.values)
plt.title('Sales Trend Over Time')
plt.xlabel('Order Date')
plt.ylabel('Sales')
plt.xticks(rotation=45)
```

```python
plt.figure(figsize=(20,5))
plt.plot(sales_by_date.index,sales_by_date.values)
plt.title('Sales Trend Over Time',fontsize=18)
plt.xlabel('Order Date',fontsize=16)
plt.ylabel('Sales',fontsize=16)
plt.xticks(rotation=45)
plt.grid(linestyle='--',color='green',alpha=0.1)
```

```python
plt.figure(figsize=(20,5))
plt.plot(sales_by_date.index,sales_by_date.values,color='red',linestyle='--')
plt.title('Sales Trend Over Time',fontsize=18)
plt.xlabel('Order Date',fontsize=16)
plt.ylabel('Sales',fontsize=16)
plt.xticks(rotation=45)
plt.grid(linestyle='--',color='green',alpha=0.8)
```

####Daily Profit Trend

```python
plt.figure(figsize=(20,5))
profit_by_date=D1.groupby('Order Date')['Profit'].sum()
plt.plot(profit_by_date.index,profit_by_date.values,color='red',linestyle='--')
plt.title('Daily profit Trend',fontsize=18)
plt.xlabel('Order Date',fontsize=16)
plt.ylabel('profit',fontsize=16)
plt.xticks(rotation=45)
plt.grid(linestyle='--',color='green',alpha=0.8)
```

```python
sales_by_region=D1.groupby('Region')['Sales'].sum()
plt.bar(sales_by_region.index,sales_by_region.values,color=['red','blue','green','orange'])
plt.title('Daily profit Trend',fontsize=18)
plt.xlabel('Region',fontsize=16)
plt.ylabel('Sales',fontsize=16)
```

```python
sales_by_region=D1.groupby('Region')['Sales'].sum()
profit_by_region=D1.groupby('Region')['Profit'].sum()
plt.bar(sales_by_region.index,sales_by_region.values,color='green')
plt.bar(profit_by_region.index,profit_by_region.values,bottom=sales_by_region.values,color='orange',label='profit')
plt.title('Total Sales by Region',fontsize=18)
plt.xlabel('Region',fontsize=16)
plt.ylabel('Sales',fontsize=16)
plt.legend()
```

```python
sales=D1.groupby('Region')['Sales'].sum()
profit=D1.groupby('Region')['Profit'].sum()

plt.figure(figsize=(12,6))
plt.bar(sales.index,sales.values,color='green')
plt.bar(profit.index,profit.values,bottom=sales.values,color='orange',label='profit')

plt.title('Sales & Profit by Region',fontsize=18)
plt.xlabel('Region',fontsize=16)
plt.ylabel('Sales',fontsize=16)
plt.legend()
```

```python
sales_by_region=D1.groupby('Region')['Sales'].sum()
plt.bar(sales_by_region.index,sales_by_region.values,color=['red','blue','green','orange'],edgecolor='yellow')
plt.title('Daily profit Trend',fontsize=18)
plt.xlabel('Region',fontsize=16)
plt.ylabel('Sales',fontsize=16)
```
