```python
"""27.Write a program that handles exceptions in a recursive factorial function."""
def factorial(n):
    if n < 0:
        raise ValueError("Negative not allowed.")
    if n == 0 or n == 1:
        return 1
    return n * factorial(n-1)

try:
    print(factorial(-3))
except Exception as e:
    print(e)
```

```python

"""43.Write a program that handles multiple custom exceptions in an e-commerce system (OutOfStockError, PaymentFailedError)."""
class OutOfStockError(Exception):
    pass

class PaymentFailedError(Exception):
    pass

def purchase(item_stock, payment_status):
    if item_stock == 0:
        raise OutOfStockError("Out of Stock!")
    if not payment_status:
        raise PaymentFailedError("Payment Failed!")

try:
    purchase(0, True)
except (OutOfStockError, PaymentFailedError) as e:
    print(e)
```
