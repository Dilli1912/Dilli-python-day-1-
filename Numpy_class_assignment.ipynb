```python
import numpy as np
# Manually created array: [Store ID, Sales, Customers, Inventory, Returns]

retail_data = np.array([
       [    1, 25000,   300,  1200,    10],
       [    2, 32000,   450,  1500,    15],
       [    3, 28000,   380,  1100,     8],
       [    4, 41000,   500,  1600,    20],
       [    5, 35000,   420,  1400,    12],
       [    6, 27000,   360,  1300,     9],
       [    7, 39000,   480,  1700,    18],
       [    8, 30000,   400,  1250,    14],
       [    9, 45000,   550,  1800,    25],
       [   10, 33000,   410,  1450,    11]
       ])
print(retail_data)
```

```python
## Operational Strategy
print("26. Which stores need **inventory restocking** based on sales and current stock?")
X=retail_data[:,1]
Y=retail_data[:,3]
S=Y<(X*0.04)   #threshold concept
print(S)
Stores_to_restock=retail_data[S,0]
print("inventory restocking stores are",Stores_to_restock)

```

```python
print("27. Which stores show **high returns percentage** and might require **quality checks**?")
X=retail_data[:,2]
Y=retail_data[:,4]
R_percen=(Y/X)*100
print(R_percen)
S=retail_data[np.argmax(R_percen),0]
print("stores show **high returns percentage** and might require **quality checks** are",S)

```

```python
print("28. Which store is the **best performer overall** (sales, customers, inventory turnover)?")
S=retail_data[:,1]
C=retail_data[:,2]
I=retail_data[:,3]
Inventory_turnover=S/I
A=((S/(S.max()))+(C/(C.max()))+(I/(I.max())))/3
print("**best performer overall** is",(np.argmax(A)+1))
```

```python
print("29. Which store is the **worst performer overall**?")
S=retail_data[:,1]
C=retail_data[:,2]
I=retail_data[:,3]
Inventory_turnover=S/I
A=((S/(S.max()))+(C/(C.max()))+(I/(I.max())))/3
print("**best performer overall** is",(np.argmin(A)+1))

```
