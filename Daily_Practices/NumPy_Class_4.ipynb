| Mode  | Meaning                  | Can Read? | Can Write? | File Created if Missing? |
|-------|--------------------------|-----------|------------|--------------------------|
| `r`   | Read text                 | ✅        | ❌         | ❌                       |
| `w`   | Write text (overwrite)    | ❌        | ✅         | ✅                       |
| `a`   | Append text               | ❌        | ✅         | ✅                       |
| `r+`  | Read + write              | ✅        | ✅         | ❌                       |
| `rb`  | Read binary               | ✅        | ❌         | ❌                       |
| `wb`  | Write binary              | ❌        | ✅         | ✅                       |


```python
with open("first.txt","w") as file:
  file.write("first")
```

```python
import PyPDF2

# Open the PDF in read-binary mode
with open("/content/Sharmil_DS_Resume (1).pdf", "rb") as file:
    reader = PyPDF2.PdfReader(file)
    print("Number of pages:", len(reader.pages))

    # Extract text from first page
    for i in range(0,len(reader.pages)):
      page = reader.pages[i]
      text = page.extract_text()
      print(f"Page {i} content:\n", text)
```

```python
import json

# Python dictionary
data = {
    "name": "John Doe",
    "age": 30,
    "skills": ["Python", "Data Engineering", "AWS"]
}

# Write to JSON file
with open("data.json", "w") as file:
    json.dump(data, file, indent=4)

print("JSON file created successfully!")
```

```python
with open("data.json", "r") as file:
    loaded_data = json.load(file)

print("Loaded JSON content:", loaded_data)
```
