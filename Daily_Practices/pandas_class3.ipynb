```python
for col in D2.select_dtypes(["int64"]).columns:
  # col_type=D2['col'].dtype
  c_min,c_max=D2[col].min(),D2[col].max()
  if str(col_type).startswith=='int':
    if c_min>np.iinfo(np.int8).min and c_max<np.iinfo(np.int8).max:
      D2[col]=D2[col].astype(np.int8)
    elif c_min>np.iinfo(np.int16).min and c_max<np.iinfo(np.int16).max:
      D2[col]=D2[col].astype(np.int16)
    elif c_min>np.iinfo(np.int32).min and c_max<np.iinfo(np.int32).max:
      D2[col]=D2[col].astype(np.int32)
    else:
      D2[col]=D2[col].astype(np.int64)
```
