```python
import PyPDF2

with open("/content/sample1.pdf","r") as file:
  reader = PyPDF2.PdfReader(file)
  print("Number of pages",len(reader.pages))
  for i in range(0,len(reader.pages)):
      page=reader.pages[i]
      text=page.extract_text()
  print(f"page {i} content:\n",text)
```
